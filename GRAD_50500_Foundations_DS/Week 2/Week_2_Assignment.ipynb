### 8. Create a plot that compares Sepal.Length to Petal.Length. Add some informative/interesting axis labels, colors, etc. Have some fun with it.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a figure with a 1x2 grid of subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Plot on the first subplot (axes[0])
sns.histplot(x="sepal length (cm)", data=iris_df, ax=axes[0], kde=True)
axes[0].set_title('Histogram of Sepal Lenght')

# Plot on the second subplot (axes[1])
sns.histplot(x="petal length (cm)", data=iris_df, ax=axes[1], kde=True)
axes[1].set_title('Histogram of Pepal Lenght')

plt.tight_layout() # Adjust layout to prevent overlapping
plt.show()
```

Now we can analyse the distribution betweeen the different species

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create a figure with a 1x2 grid of subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Plot on the first subplot (axes[0])
sns.histplot(x="sepal length (cm)", data=iris_df, ax=axes[0], hue='target_name', kde=True)
axes[0].set_title('Histogram of Sepal Lenght')

# Plot on the second subplot (axes[1])
sns.histplot(x="petal length (cm)", data=iris_df, ax=axes[1], hue='target_name', kde=True)
axes[1].set_title('Histogram of Pepal Lenght')

plt.tight_layout() # Adjust layout to prevent overlapping
plt.show()
```
