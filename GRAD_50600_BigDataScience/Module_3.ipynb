## Part 1: Example Attempted

In this project, I will choose the different visualizations from the **"Big Data and Visualization Examples"** with the purpose to explore practice how to handle larger samples of data, instead of traditional visualizations:

First we will do the example of "small" data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns


data = {
        "Year":[2018,2019,2020,2021,2022],
        "Sales":[100,150,200,250,300]}

df = pd.DataFrame(data)

#Plotting data

plt.figure(figsize=(10,5))
plt.plot(df['Year'], df['Sales'], marker = "o")
plt.title("Sales Over Years (Traditional Visualization)")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

Now I will do the big data example, only that I would make a couple of modifications

```python
import dask.dataframe as dd

rows = 12**8

larger_data = pd.DataFrame({
        "Year": np.random.randint(2000,2023,size=rows),
        "Sales":np.random.randint(100,1000,size=rows)})

larger_data.to_parquet("data/large_dataset.parquet")
```

```python
larger_data = dd.read_parquet("data/large_dataset.parquet")

agg_data = larger_data.groupby("Year")['Sales'].mean().compute()

#Plotting data

plt.figure(figsize=(10,5))
agg_data.plot(kind="bar")
plt.title("Average sales per year (Big Data Visualization)")
plt.xlabel("Year")
plt.ylabel("Average Sales")
plt.grid(True)
plt.show()
```
